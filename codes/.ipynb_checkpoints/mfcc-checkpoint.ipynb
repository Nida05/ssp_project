```python
def get_ravdess_info(filename):
    parts = filename.split('-')
    
    modality = int(parts[0])
    channel = int(parts[1])
    emotion = int(parts[2])
    actor = int(parts[6].split('.')[0])
    
    return modality, channel, emotion, actor
```

```python
def map_emotion(emotion_id):
    mapping = {
        1: 'neutral',
        2: 'neutral',   # calm merged
        3: 'happy',
        4: 'sad',
        5: 'angry',
        6: 'fear',
        7: 'disgust',
        8: 'neutral'    # surprise merged
    }
    return mapping[emotion_id]
```

```python
import os

data_path = "path_to_RAVDESS"

actors = [f"Actor_{i:02d}" for i in range(1, 11)]

X = []
y = []

for actor in actors:
    actor_path = os.path.join(data_path, actor)
    
    for file in os.listdir(actor_path):
        if file.endswith(".wav"):
            
            modality, channel, emotion_id, actor_id = get_ravdess_info(file)
            
            if modality != 3 or channel != 1:
                continue
            
            label = map_emotion(emotion_id)
            
            file_path = os.path.join(actor_path, file)
            mfcc = extract_mfcc(file_path)
            
            X.append(mfcc)
            y.append(label)
```
